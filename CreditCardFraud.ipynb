```python
from sklearn.model_selection import train_test_split

X = df.drop('Class', axis=1)
y = df['Class']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

X_train['scaled_amount'] = scaler.fit_transform(X_train[['Amount']])
X_test['scaled_amount'] = scaler.transform(X_test[['Amount']])

X_train = X_train.drop('Amount', axis=1)
X_test = X_test.drop('Amount', axis=1)
```

```python
print(X_train.columns.tolist())
```

```python
X_train = X_train.drop('Time', axis=1)
X_test = X_test.drop('Time', axis=1)
```

```python
print(X_train.columns.tolist())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

log_model = LogisticRegression(max_iter=1000, random_state=42)
log_model.fit(X_train, y_train)

y_pred = log_model.predict(X_test)

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
log_model_bal = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42)
log_model_bal.fit(X_train, y_train)
y_pred_bal = log_model_bal.predict(X_test)
print(classification_report(y_test, y_pred_bal))
```

```python
print(confusion_matrix(y_test, y_pred_bal))
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)

print(y_train_sm.value_counts())
```

```python
log_model_sm = LogisticRegression(max_iter=1000, random_state=42)
log_model_sm.fit(X_train_sm, y_train_sm)

y_pred_sm = log_model_sm.predict(X_test)
print(classification_report(y_test, y_pred_sm))
print(confusion_matrix(y_test, y_pred_sm))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train_sm, y_train_sm)

y_pred_rf = rf_model.predict(X_test)
print(classification_report(y_test, y_pred_rf))
print(confusion_matrix(y_test, y_pred_rf))
```

```python
importances = pd.Series(rf_model.feature_importances_, index=X_train.columns)
top_features = importances.sort_values(ascending=False).head(10)

print(top_features)
```

```python
top_features.plot(kind='barh')
plt.title('Top 10 Important Features - Random Forest')
plt.xlabel('Importance')
plt.gca().invert_yaxis()  
plt.show()
```

```python
sns.boxplot(x='Class', y='V14', data=df)
plt.title('V14 Distribution by Class')
plt.show()
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score

models = {
    'LogReg baseline': log_model,
    'LogReg class_weight': log_model_bal,
    'LogReg SMOTE': log_model_sm,
    'RF SMOTE': rf_model
}

for name, model in models.items():
    probs = model.predict_proba(X_test)[:, 1]
    roc_auc = roc_auc_score(y_test, probs)
    pr_auc = average_precision_score(y_test, probs)
    print(f"{name}: ROC-AUC = {roc_auc:.4f}, PR-AUC = {pr_auc:.4f}")
```
